## x축 -5에서 5까지의 그래프 $y = x^2$

- `0.1`을 더 작게 하면 곡선이 더 부드러워짐 (numpy의 `arange()` 활용)
- `ls`는 `linestyle`과 동일, `":"`은 점선(dotted line)
- 4가지 기본 스타일: solid(`-`), dotted(`:`), dashed(`--`), dashdot(`-.`), None(`''`, `' '`)
- 레이블에서 앞뒤의 `$`은 내용을 LaTeX 형식의 수식으로 표현

```python
import numpy as np

x = np.arange(-5, 5.1, 0.1)   # x축에서 이웃한 좌표의 거리를 0.1로 지정
plt.plot(x, x**2, ls=":");
plt.xlabel('$x$')             # 레이블을 LaTeX 수식으로 표현
plt.ylabel('$y = x^2$');
```

## 함수 `plot()`의 패러미터 `linestyle`과 `marker`

지정 가능한 선 종류 4가지:

| 선 스타일 이름 | 기호 | 예제 |
|---|---|---|
| 실선 (Solid line) | `-` | `plt.plot(x, y, linestyle='-')` |
| 점선 (Dashed line) | `--` | `plt.plot(x, y, linestyle='--')` |
| 점선 (Dotted line) | `:` | `plt.plot(x, y, linestyle=':')` |
| 대시 점선 (Dash-dot line) | `-.` | `plt.plot(x, y, linestyle='-.')` |

### 다양한 선 종류(linestyle) 결과

```python
import matplotlib.pyplot as plt

# 데이터 준비
x = [1, 2, 3, 4]
y1 = [1, 4, 9, 16]
y2 = [2, 4, 6, 8]
y3 = [3, 6, 9, 12]
y4 = [4, 8, 12, 16]

# 선 스타일별로 그래프 그리기
plt.plot(x, y1, linestyle='-', label='Solid line')      # 실선
plt.plot(x, y2, linestyle='--', label='Dashed line')    # 점선
plt.plot(x, y3, linestyle=':', label='Dotted line')     # 도트 점선
plt.plot(x, y4, linestyle='-.', label='Dash-dot line')  # 대시 도트 점선

# 그래프 제목과 축 레이블
plt.title("선 종류(Line Styles in Matplotlib)")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

# 범례 추가
plt.legend()
# 그래프 보여주기
plt.show()
```

## 마커 종류

마커(marker)를 사용하여 각 데이터 포인트를 강조합니다.

| 마커 이름 | 기호 | | 마커 이름 | 기호 |
|---|---|---|---|---|
| 점 (Point) | `.` | | 사각형 (Square) | `s` |
| 픽셀 (Pixel) | `,` | | 다이아몬드 (Diamond) | `D` |
| 원 (Circle) | `o` | | 작은 다이아몬드 (Thin diamond) | `d` |
| 삼각형 (Triangle_down) | `v` | | 육각형1 (Hexagon1) | `h` |
| 역삼각형 (Triangle_up) | `^` | | 육각형2 (Hexagon2) | `H` |
| 오른쪽 삼각형 (Triangle_right) | `>` | | 별 (Star) | `*` |
| 왼쪽 삼각형 (Triangle_left) | `<` | | 십자 (Plus) | `+` |
| 곱하기 (X) | `x` | | 작은 십자 (Thin plus) | `\|` |
| 작은 곱하기 (Thin x) | `_` | | | |

### 5개 마커 사용

```python
import matplotlib.pyplot as plt

# 데이터 준비
x = [1, 2, 3, 4]
y1 = [1, 4, 9, 16]
y2 = [2, 3, 4, 5]
y3 = [2, 2, 3, 3]
y4 = [3, 1, 4, 2]
y5 = [4, 5, 6, 8]

# 그래프 그리기
plt.plot(x, y1, marker='o', label='Circle')        # 원
plt.plot(x, y2, marker='^', label='Triangle_up')   # 삼각형 위
plt.plot(x, y3, marker='s', label='Square')        # 사각형
plt.plot(x, y4, marker='*', label='Star')          # 별
plt.plot(x, y5, marker=',', label='Pixel')         # 픽셀

# 그래프 제목과 축 레이블
plt.title("Marker Styles in Matplotlib")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
# 범례 추가
plt.legend()
# 그래프 보여주기
plt.show()
```

## 빈도 수 값을 막대 그래프에 표시

함수 `sns.countplot()`의 반환 객체는 `Axes` 객체입니다.

- `ax.patches`로 각 막대(patch)에 대해 반복문 실행
- `p.get_height()`로 현재 막대의 높이(빈도 수)를 가져옴
- `ax.text(x, y, s=int(height), ...)`로 막대 위 위치에 빈도수 표시

`Axes.text(x, y, s, fontdict=None, **kwargs)` 인자 설명:
- `x = (p.get_x() + p.get_width()/2.)`: 텍스트를 막대의 중앙에 위치
- `y = (height + .2)`: 텍스트를 막대 바로 위에 위치
- `s = int(height)`: 텍스트 내용은 막대 높이(빈도 수)
- `ha = 'center'`: 중앙 정렬 (미지정 시 왼쪽 정렬되어 약간 오른쪽으로 표시)
- `size = 12`: 텍스트 크기, `color = 'blue'`: 텍스트 색상

```python
import seaborn as sns

ax = sns.countplot(data=mtc, x='gear',
                   order=mtc.gear.value_counts().index)

for p in ax.patches:
    height = p.get_height()  # 빈도 수
    ax.text(x=(p.get_x() + p.get_width() / 2.), y=(height + .2),
            s=int(height), ha='center', size=12, color='blue')

ax.set_ylim(0, 17)  # y축의 범위를 0부터 17까지로 설정
plt.show()
```

## 상관계수 (correlation coefficient)

상관관계 분석에서 두 변수 간 선형 관계의 정도를 수량화한 값입니다.

- 상관계수 r은 `[-1, 1]` 실수 값
- r이 0에 근접할수록 선형 관계가 약해짐
- 양수 r: 양의 상관관계 (두 변수 값이 함께 증가)
- 음수 r: 음의 상관관계 (한 변수가 증가할 때 다른 변수는 감소)

## 유명한 붓꽃 데이터 `iris`를 seaborn에서 가져오기

seaborn의 `load_dataset("iris")`는 열 이름이 `sepal_length`, `sepal_width`, `petal_length`, `petal_width`, `species` 형식입니다.

## `df`의 메소드 `df.corr()`

열들 간의 상관관계 데이터프레임을 반환합니다. 대각선은 1이며 대각선을 기준으로 대칭입니다.
꽃잎 너비(`petal_width`)와 꽃잎 길이(`petal_length`)의 상관관계는 0.963으로 매우 높습니다.

```python
# 상관관계 계수 계산
import pandas as pd

pd.set_option('display.precision', 3)
corr_iris = iris[iris.columns[:-1].to_list()].corr()
corr_iris
```

## 상관관계 계수를 색상으로 표현 — 히트맵(heatmap)

값에 따라 색깔이 변하는 행렬 그림입니다. 상관계수가 1에 가까우면 밝아지고 -1에 가까우면 어두워지므로,
상관계수가 1인 대각선이 가장 밝습니다. `annot=True`로 값을 함께 표시합니다.

```python
sns.heatmap(corr_iris, annot=True);
```

## 히트맵을 보기 좋게 좌측 하단의 삼각 행렬만 남도록 수정하기 위한 준비

- 마스크(mask): `corr_iris`와 같은 모양(4×4)이 모두 0으로 채워진 2차원 행렬 생성
- `np.triu_indices_from(mask)`: mask의 상삼각형에 대한 인덱스 반환
- 오른쪽 위 대각 행렬을 모두 1로 수정
- `mask[1:, :-1]`: 첫 번째 행과 마지막 열을 제거

```python
# mask 만들기
import numpy as np

mask = np.zeros_like(corr_iris)
mask
```

```python
np.triu_indices_from(mask)
```

```python
# 오른쪽 위 대각 행렬을 1로 바꾸기
mask[np.triu_indices_from(mask)] = 1
mask
```

```python
mask_new = mask[1:, :-1]  # mask 첫 번째 행, 마지막 열 제거
mask_new
```

## 히트맵의 좌측 하단의 삼각 행렬 그리기

상관관계 계수 데이터프레임의 첫 번째 행과 마지막 열을 제거한 `corr_iris_new`를 생성합니다.

- `mask=mask_new`
- `cmap='coolwarm'`: 양의 상관관계는 붉은색, 음의 상관관계는 파란색
- `cbar_kws={'shrink': .4}`: 컬러바 줄이기
- `linewidths=.5`: 경계 구분선 추가

```python
corr_iris_new = corr_iris.iloc[1:, :-1]

sns.heatmap(corr_iris_new, annot=True, cmap='coolwarm', mask=mask_new,
            annot_kws={"size": 10}, vmin=-1, vmax=1,
            cbar_kws={'shrink': .4},  # 범례 크기 줄이기
            linewidths=.5);           # 경계 구분선 추가
```

## 결측 값이 있는 행 출력 — 함수 `isnull()`

`display_side_by_side`는 두 데이터프레임을 좌우로 나란히 표시하는 보조 함수입니다.
(주피터 환경에서 HTML로 출력)

```python
# 두 데이터프레임을 좌우로 나란히 표시하는 보조 함수
from IPython.display import display_html

def display_side_by_side(*args):
    html_str = ''
    for df_ in args:
        html_str += df_.to_html()
    display_html(
        html_str.replace('table', 'table style="display:inline-block; margin-right:20px;"'),
        raw=True
    )
```

```python
display_side_by_side(df[df.Ozone.isnull()], df[df['Solar.R'].isnull()])
```

## 함수 `DataFrame.groupby()`와 `DataFrameGroupBy.transform()`

- **groupby() + 집계 함수**: 그룹 수만큼 행이 줄어들어 요약 테이블 (그룹별 합계·평균 리포트에 적합)
- **groupby() + transform()**: 원본과 동일한 행 수를 유지하며, 각 행에 자신이 속한 그룹의 집계값을 채워 줌

`transform()`은 각 그룹에 대해 동일한 인덱스의 DataFrame을 생성하는 함수를 호출하여,
원래 객체와 동일한 인덱스를 가진 DataFrame을 반환합니다.

## 월별 오존의 평균 계산

- `df.groupby('Month')['Ozone'].mean()`: Month별 Ozone 값의 평균
- `df.groupby('Month')['Ozone'].transform('mean')`: 월별 Ozone 평균값을 원본 각 행에 반복해서 붙임

```python
f1 = df.groupby('Month')['Ozone'].mean().to_frame()
f2 = df.groupby('Month')['Ozone'].transform('mean').to_frame()
```

```python
display_side_by_side(f1, f2)
```

## 결측값 처리

전략:
- **Ozone**: 월별 평균값으로 대체
- **Solar.R**: 전체 평균값으로 대체

```python
# 2. 결측값 처리 전략
# Ozone: 월별 평균값으로 대체
ozone_monthly_mean = df.groupby('Month')['Ozone'].transform('mean')
df['Ozone'] = df['Ozone'].fillna(ozone_monthly_mean)

# Solar.R: 전체 평균값으로 대체
df['Solar.R'] = df['Solar.R'].fillna(df['Solar.R'].mean())

print("\n--- 결측값 처리 후 ---")
print(df.isnull().sum())
```
