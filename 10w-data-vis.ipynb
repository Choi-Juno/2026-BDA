```python
x = [1, 2, 3, 4]
y1 = [1, 4, 9, 16]
y2 = [2, 4, 6, 8]
y3 = [3, 6, 9, 12]
y4 = [4, 8, 12, 16]

plt.plot(x, y1, linestyle='-', label='Solid line')
plt.plot(x, y2, linestyle='--', label='Dashed line')
plt.plot(x, y3, linestyle=':', label='Dotted line')
plt.plot(x, y4, linestyle='-.', label='Dash-dot line')

plt.title("선 종류")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```

```python
# 데이터 준비
x = [1, 2, 3, 4]
y1 = [1, 4, 9, 16]
y2 = [2, 3, 4, 5]
y3 = [2, 2, 3, 3]
y4 = [3, 1, 4, 2]
y5 = [4, 5, 6, 8]
# 그래프 그리기
plt.plot(x, y1, marker='o', label='Circle')      # 원
plt.plot(x, y2, marker='^', label='Triangle_up') # 삼각형 위
plt.plot(x, y3, marker='s', label='Square')      # 사각형
plt.plot(x, y4, marker='*', label='Star')        # 별
plt.plot(x, y5, marker=',', label='Pixel')       # 픽셀
# 그래프 제목과 축 레이블
plt.title("Marker Styles in Matplotlib")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
# 범례 추가
plt.legend()
# 그래프 보여주기
plt.show()
```

```python
import pandas as pd
pd.set_option('display.precision', 3)
corr_iris = iris[iris.columns[:-1].to_list()].corr()
corr_iris
```

```python
# 히트맵을 보기 좋게 좌측 하단의 삼각 행렬만 남도록 수정하기 위한 준비
import numpy as np

mask = np.zeros_like(corr_iris)
mask
```

```python
np.triu_indices_from(mask) # mask의 상삼각형에 대한 인덱스를 만환
```

```python
mask[np.triu_indices_from(mask)] = 1
mask
```

```python
mask_new = mask[1:, :-1]
mask_new
```

```python
corr_iris_new = corr_iris.iloc[1:, :-1]
corr_iris_new
```

```python
sns.heatmap(corr_iris_new, annot=True, cmap='coolwarm', mask=mask_new, annot_kws={'size': 10}, vmin=-1, vmax=1, cbar_kws={'shrink': .4}, linewidth=.5)
```
