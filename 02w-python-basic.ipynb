### Set

```python
basket = {'apple', 'orange', 'apple', 'pear'}
print(basket)
```

Set은 mutable한 객체이나, Set의 원소는 immutable한 객체만 들어갈 수 있다.

```python
s1 = set([1, 2, 3, 4, 5])
s2 = set([4, 5, 6, 7, 8])
s1.union(s2)
s1.intersection(s2)
s1.difference(s2)
s1.symmetric_difference(s2)
```

```python
# s1.add([0, 1])
# list는 mutable한 객체이므로 추가할 수 없다.
```
